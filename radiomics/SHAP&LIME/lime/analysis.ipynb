```python
import pandas as pd
from scipy import stats
```

```python
test1_data = pd.read_csv('/home/pyq6817/IPMN-Radiomics-Plus-Deeplearning/radiomics/SHAP&LIME/lime/2D/2D_Train_Test_1/Test1_lime_values.csv')
test2_data = pd.read_csv('/home/pyq6817/IPMN-Radiomics-Plus-Deeplearning/radiomics/SHAP&LIME/lime/2D/2D_Train_Test_2/Test2_lime_values.csv')
test3_data = pd.read_csv('/home/pyq6817/IPMN-Radiomics-Plus-Deeplearning/radiomics/SHAP&LIME/lime/2D/2D_Train_Test_3/Test3_lime_values.csv')
test4_data = pd.read_csv('/home/pyq6817/IPMN-Radiomics-Plus-Deeplearning/radiomics/SHAP&LIME/lime/2D/2D_Train_Test_4/Test4_lime_values.csv')
```

```python
test3_data = test3_data.drop(columns=['True_Label'])
feature_cols = [col for col in test3_data.columns if col not in ['Case_Name', 'Fold']]
test3_data[feature_cols] = test3_data[feature_cols].abs()
# select from fold1 to fold5, if Fold = 1 , select this line
test3_data_1 = test3_data[test3_data['Fold'] == 1]
test3_data_2 = test3_data[test3_data['Fold'] == 2]
test3_data_3 = test3_data[test3_data['Fold'] == 3]
test3_data_4 = test3_data[test3_data['Fold'] == 4]
test3_data_5 = test3_data[test3_data['Fold'] == 5]
# fold level average
test3_data_folds = pd.concat([test3_data_1, test3_data_2, test3_data_3, test3_data_4, test3_data_5])
test3_data_avg = test3_data_folds.groupby('Case_Name').mean().reset_index()
test3_data_avg = test3_data_avg.drop(columns=['Fold'])
test3_data_avg['total-lime'] = test3_data_avg.drop(columns=['Case_Name']).sum(axis=1)
test3_data_avg = test3_data_avg[['Case_Name', 'total-lime']]
```

```python
test4_data = test4_data.drop(columns=['True_Label'])
feature_cols = [col for col in test4_data.columns if col not in ['Case_Name', 'Fold']]
test4_data[feature_cols] = test4_data[feature_cols].abs()
# select from fold1 to fold5, if Fold = 1 , select this line
test4_data_1 = test4_data[test4_data['Fold'] == 1]
test4_data_2 = test4_data[test4_data['Fold'] == 2]
test4_data_3 = test4_data[test4_data['Fold'] == 3]
test4_data_4 = test4_data[test4_data['Fold'] == 4]
test4_data_5 = test4_data[test4_data['Fold'] == 5]
# fold level average
test4_data_folds = pd.concat([test4_data_1, test4_data_2, test4_data_3, test4_data_4, test4_data_5])
test4_data_avg = test4_data_folds.groupby('Case_Name').mean().reset_index()
test4_data_avg = test4_data_avg.drop(columns=['Fold'])
test4_data_avg['total-lime'] = test4_data_avg.drop(columns=['Case_Name']).sum(axis=1)
test4_data_avg = test4_data_avg[['Case_Name', 'total-lime']]
```

```python
# merge all lime values
total_lime_values = pd.concat([test1_data_avg, test2_data_avg, test3_data_avg, test4_data_avg])
total_lime_values
```

```python
uncertainty_data = pd.read_excel('/home/pyq6817/IPMN-Radiomics-Plus-Deeplearning/radiomics/SHAP&LIME/IPMN_all_cases_final.xlsx')
uncertainty_data = uncertainty_data[['ID', 'Total_uncertainty']]

uncertainty_data.rename(columns={'ID': 'Case_Name'}, inplace=True)
uncertainty_data
# merge uncertainty with lime values
total_lime_values_with_uncertainty = pd.merge(total_lime_values, uncertainty_data, on='Case_Name', how='left')
```
